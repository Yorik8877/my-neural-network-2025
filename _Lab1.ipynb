```python
import math
```

```python
f_sig = lambda x: 1/(1 + math.exp(-1*x))
```

```python
anti_f_sig = lambda x: (1 - x)*x
```

```python
O_ideal = 1
O_ideal_sig = f_sig(O_ideal)
print('O_ideal:', O_ideal)
print('O_ideal_sig:', O_ideal_sig)
```

```python
E = 0.7
a = 0.3
```

```python
calc_dw = lambda grad,old_dw: E*grad + a*old_dw
```

```python
a,b,c = 3,5,2
```

```python
w1, w2, w3 = 0.1, 0.4, 0.3
w4, w5, w6 = 0.9, 0.7, 0.2
w7, w8, w9 = 0.31, 0.11, 0.27
w10, w11, w12 = 0.47, 0.51, 0.67

dw1, dw2, dw3 = 0, 0, 0
dw4, dw5, dw6 = 0, 0, 0
dw7, dw8, dw9 = 0, 0, 0
dw10, dw11, dw12 = 0, 0, 0

for i in range(2):
    print(f'\n=== Шаг {i+1} ===')
    
    print(f'w1, w2, w3 = {w1, w2, w3}')
    print(f'w4, w5, w6 = {w4, w5, w6}')
    print(f'w7, w8, w9 = {w7, w8, w9}')
    print(f'w10, w11, w12 = {w10, w11, w12}')
    print()
    
    H1_in = a*w1 + b*w4 + c*w7
    H2_in = a*w2 + b*w5 + c*w8
    H3_in = a*w3 + b*w6 + c*w9
    
    print()
    print(f'H1_in:\t{H1_in};\t\t\t H2_in: {H2_in};\t\t\t H3_in: {H3_in}')
    
    H1_out = f_sig(H1_in)
    H2_out = f_sig(H2_in)
    H3_out = f_sig(H3_in)
    
    print(f'H1_out:\t{H1_out};\t H2_out: {H2_out};\t H3_out: {H3_out}')
    
    O_in = H1_out*w10 + H2_out*w11 + H3_out*w12
    O_out = f_sig(O_in)
    
    print()
    print(f'O_in:\t{O_in};\t O_out: {O_out}')
    
    dO = (O_ideal_sig - O_out)*anti_f_sig(O_out)
    
    print()
    print(f'dO:\t{dO}')
    
    dH1 = anti_f_sig(O_out)*dO*w10
    dH2 = anti_f_sig(O_out)*dO*w11
    dH3 = anti_f_sig(O_out)*dO*w12
    
    print(f'dH1:\t{dH1};\t dH2: {dH2};\t dH3: {dH3}')
    print()
    
    grad_w1 = a*dH1
    grad_w2 = a*dH2
    grad_w3 = a*dH3
    
    print(f'grad_w1, grad_w2, grad_w3 = {grad_w1, grad_w2, grad_w3}')
    
    grad_w4 = b*dH1
    grad_w5 = b*dH2
    grad_w6 = b*dH3
    
    print(f'grad_w4, grad_w5, grad_w6 = {grad_w4, grad_w5, grad_w6}')
    
    grad_w7 = c*dH1
    grad_w8 = c*dH2
    grad_w9 = c*dH3
    
    print(f'grad_w7, grad_w8, grad_w9 = {grad_w7, grad_w8, grad_w9}')
    
    grad_w10 = dO*dH1
    grad_w11 = dO*dH2
    grad_w12 = dO*dH3
    
    print(f'grad_w10, grad_w11, grad_w12 = {grad_w10, grad_w11, grad_w12}')
    
    dw1, dw2, dw3 = calc_dw(grad_w1, dw1), calc_dw(grad_w2, dw2), calc_dw(grad_w3, dw3)
    dw4, dw5, dw6 = calc_dw(grad_w4, dw4), calc_dw(grad_w5, dw5), calc_dw(grad_w6, dw6)
    dw7, dw8, dw9 = calc_dw(grad_w7, dw7), calc_dw(grad_w8, dw8), calc_dw(grad_w9, dw9)
    dw10, dw11, dw12 = calc_dw(grad_w10, dw10), calc_dw(grad_w11, dw11), calc_dw(grad_w12, dw12)
    
    print()
    print(f'dw1, dw2, dw3 = {dw1, dw2, dw3}')
    print(f'dw4, dw5, dw6 = {dw4, dw5, dw6}')
    print(f'dw7, dw8, dw9 = {dw7, dw8, dw9}')
    print(f'dw10, dw11, dw12 = {dw10, dw11, dw12}')
    
    w1, w2, w3 = w1 + dw1, w1 + dw1, w1 + dw1
    w4, w5, w6 = w4 + dw4, w5 + dw5, w6 + dw6
    w7, w8, w9 = w7 + dw7, w8 + dw8, w9 + dw9
    w10, w11, w12 = w10 + dw10, w11 + dw11, w12 + dw12
    
    
```
