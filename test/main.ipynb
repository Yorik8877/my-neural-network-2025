```python
import math
import csv
```

```python
dataset = [];

# Чтение датасета
with open('test_data.csv', 'r') as csvDataset:
    csvDatasetReader = csv.reader(csvDataset, delimiter='\t')
    for row in csvDatasetReader:
        dataset.append([ float(row[0]), float(row[1]), float(row[2]) ])
```

```python
# Сигмоид
f_sig = lambda x: 1/(1 + math.exp(-1*x))
# Конечно-разностный аналог сигмоида для метода обр распределения
anti_f_sig = lambda x: (1 - x)*x
```

```python
O_IDEAL = 1
O_IDEAL_SIG = f_sig(O_IDEAL)
print('O_IDEAL:', O_IDEAL)
print('O_IDEAL_SIG:', O_IDEAL_SIG)
```

```python
E = 0.7
a = 0.3
```

```python
# Рассчитать дельту
calc_dw = lambda grad,old_dw: E*grad + a*old_dw
```

```python
a,b,c = 3,5,2
```

```python
calc_H_in = lambda w1,w2,w3: a*w1 + b*w2 + c*w3
calc_O_in = lambda a,b,c,w1,w2,w3: a*w1 + b*w2 + c*w3
```

```python
def calculateMse(answerList: list, ideal:float, real: float):
    mse = float(0)
    for answer in answerList:
        mse += ((ideal - real)**2)
    
    return mse / len(answerList)
```
