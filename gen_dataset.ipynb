```python
import math
import csv
import random

def generate_discriminant_data(num_samples=1000, a_range=(0, 5), b_range=(0, 5), c_range=(0, 5)):
    """
    Генерирует данные для дискриминанта квадратного уравнения
    Формула: D = b² - 4ac
    """
    data = []
    
    for _ in range(num_samples):
        # Генерируем случайные дробные значения
        a = random.uniform(a_range[0], a_range[1])
        b = random.uniform(b_range[0], b_range[1]) 
        c = random.uniform(c_range[0], c_range[1])
        
        # Вычисляем дискриминант
        discriminant = b**2 - 4*a*c
        
        # Добавляем в данные
        data.append([a, b, c, discriminant])
    
    return data

def save_to_csv(data, filename="discriminant_data.csv"):
    """Сохраняет данные в CSV файл"""
    with open(filename, 'w', newline='') as csvfile:
        writer = csv.writer(csvfile, delimiter='\t')
        writer.writerow(['a', 'b', 'c', 'discriminant'])  # заголовок
        for row in data:
            writer.writerow(row)
    print(f"Данные сохранены в {filename}")

def analyze_data(data):
    """Анализирует сгенерированные данные"""
    discriminants = [row[3] for row in data]
    
    print("Анализ сгенерированных данных:")
    print(f"Количество образцов: {len(data)}")
    print(f"Диапазон a: [{min(row[0] for row in data):.2f}, {max(row[0] for row in data):.2f}]")
    print(f"Диапазон b: [{min(row[1] for row in data):.2f}, {max(row[1] for row in data):.2f}]")
    print(f"Диапазон c: [{min(row[2] for row in data):.2f}, {max(row[2] for row in data):.2f}]")
    print(f"Диапазон дискриминанта: [{min(discriminants):.2f}, {max(discriminants):.2f}]")
    print(f"Среднее значение дискриминанта: {sum(discriminants)/len(discriminants):.2f}")
    
    # Анализ знаков дискриминанта
    positive = sum(1 for d in discriminants if d > 0)
    zero = sum(1 for d in discriminants if d == 0)
    negative = sum(1 for d in discriminants if d < 0)
    
    print(f"Положительные дискриминанты: {positive} ({positive/len(data)*100:.1f}%)")
    print(f"Нулевые дискриминанты: {zero} ({zero/len(data)*100:.1f}%)")
    print(f"Отрицательные дискриминанты: {negative} ({negative/len(data)*100:.1f}%)")

# Генерируем данные
print("Генерация данных для дискриминанта...")
data = generate_discriminant_data(1000)

# Анализируем данные
analyze_data(data)

# Сохраняем в файл
save_to_csv(data, "discriminant_data.csv")

# Выводим первые 10 примеров для проверки
print("\nПервые 10 примеров:")
print("a\tb\tc\tD")
for i in range(min(10, len(data))):
    a, b, c, d = data[i]
    print(f"{a:.3f}\t{b:.3f}\t{c:.3f}\t{d:.3f}")

# Дополнительная версия с нормализацией данных
def generate_normalized_discriminant_data(num_samples=1000):
    """
    Генерирует данные с нормализованным дискриминантом
    """
    data = []
    max_discriminant = 25  # максимально возможное значение при a,b,c ∈ [0,5]
    
    for _ in range(num_samples):
        a = random.uniform(0, 5)
        b = random.uniform(0, 5)
        c = random.uniform(0, 5)
        
        discriminant = b**2 - 4*a*c
        # Нормализуем дискриминант в диапазон [0, 1]
        normalized_discriminant = (discriminant + 100) / 125  # смещаем чтобы убрать отрицательные значения
        
        data.append([a, b, c, normalized_discriminant])
    
    return data

print("\n" + "="*50)
print("Генерация нормализованных данных...")
normalized_data = generate_normalized_discriminant_data(1000)
save_to_csv(normalized_data, "discriminant_data_normalized.csv")

# Анализ нормализованных данных
discriminants_norm = [row[3] for row in normalized_data]
print(f"Диапазон нормализованного дискриминанта: [{min(discriminants_norm):.3f}, {max(discriminants_norm):.3f}]")

print("\nПервые 10 нормализованных примеров:")
print("a\tb\tc\tD_norm")
for i in range(min(10, len(normalized_data))):
    a, b, c, d_norm = normalized_data[i]
    print(f"{a:.3f}\t{b:.3f}\t{c:.3f}\t{d_norm:.3f}")
```
