```python
import random
import numpy as np

# Размер изображения
INPUT_DIM = 40*40
# Классификация: кружки или квадраты
OUTPUT_DIM = 2

# Скрытые нейроны
H_DIM = 3

# Скорость обучения
ALPHA = 0.01

# TODO: заменить на чтение изображения
x = np.random.randn(1, INPUT_DIM)
y = random.randint(0, OUTPUT_DIM-1)

# Матрица весов между входными и скрытыми нейронами
W1 = np.random.randn(INPUT_DIM, H_DIM)
# Биас между входными и скрытыми нейронами
B1 = np.random.randn(1, H_DIM)

# Матрица весов между входными и скрытыми нейронами
W2 = np.random.randn(H_DIM, OUTPUT_DIM)
# Биас между входными и скрытыми нейронами
B2 = np.random.randn(1, OUTPUT_DIM)
```

```python
def relu(x):
    return np.maximum(x,0)

# t - вектор
def relu_deriv(t):
    return (t >= 0).astype(float)

def softmax(t):
    out = np.exp(t)
    return out / np.sum(out)

def sparse_cross_entropy(z,y):
    return -np.log(z[0, y])

def to_full(y, num_classes):
    y_full = np.zeros((1, num_classes))
    y_full[0, y] = 1
    return y_full
```

```python
# Прямое распространение
# Рассчитываем вектор сумм H_in
t1 = x @ W1 + B1
# Рассчитываем вектор H_out
h1 = relu(t1)

# Рассчитываем выходное значение O
t2 = h1 @ W2 + B2
# Вектор вероятностей
z = softmax(t2)

# Ошибка 
E = sparse_cross_entropy(z, y)

# Обратное распространение
y_full = to_full(y, OUTPUT_DIM)
dE_dt2 = z - y_full
dE_dW2 = h1.T @ dE_dt2
dE_dB2 = dE_dt2

dE_dh1 = dE_dt2 @ W2.T
dE_dt1 = dE_dh1 * relu_deriv(t1)
dE_dW1 = x.T @ dE_dt1
dE_dB1 = dE_dt1

# Обновление весов
# TODO: привести к нашему виду
W1 = W1 - ALPHA * dE_dW1
B1 = B1 - ALPHA * dE_dB1
W2 = W2 - ALPHA * dE_dW2
B2 = B2 - ALPHA * dE_dB2
```
